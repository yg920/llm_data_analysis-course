```python
#컴렁 추가(파생 컬럼)
#라인 합계 금액 : line_total
order_items["line_total"] = order_items["quantity"] * order_items["unit_price"]

print(order_items.head(10))

print(order_items.info())
```

```python
print(orders["order_id"].duplicated().sum())
```

```python
# print(order_items.head(10))
# order_status를 함께 출력 하려면 => merge

order_sales = order_items.merge(
    orders[["order_id", "customer_id", "order_date", "order_status"]],
      on="order_id", how="left", validate="many_to_one"
)

print(order_sales.head())
```

```python
order_sales_test2 = order_sales_test.copy()

print(order_sales_test2.head())
```

```python
order_sales_test2.loc[0, "order_date"] = pd.Timestamp("2026-02-28")
```

```python
order_sales_test2["order_date"] = pd.to_datetime(
    order_sales_test2["order_date"], 
    errors="coerce"
)

print(order_sales_test2.info())
print(order_sales_test2.head(5))
print(order_sales_test2["order_date"].isna().sum())
```
